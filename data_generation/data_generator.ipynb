```python
import numpy as np
import pandas as pd
np.random.seed(23)
stops = [f"S{i}" for i in range(1,51)]
pg_t=["R","S","SC"]#R-Regular S-Student SC-Senior Citizen
tp={"R":50,"S":20,"SC":25}

date_range=pd.date_range("2025-01-01","2025-09-04",freq="D")
data=[]
t=1
for date in date_range:
    for i in range(50):
        entry=np.random.choice(stops)
        exit= np.random.choice(stops)
        while (exit==entry):
            entry=np.random.choice(stops)
            exit= np.random.choice(stops)
        pg=np.random.choice(pg_t)
        peak_hour = np.random.choice([True, False])
        ticket=0
        ns=abs(int(exit[1:])-int(entry[1:]))
        if((peak_hour)):
            ticket+=tp[pg]*ns*1.2
        else:
            ticket+=tp[pg]*ns
        duration=ns*5
        data.append([t,date,entry,exit,pg,duration,peak_hour,round(ticket,2)])
        t+=1

df = pd.DataFrame(data, columns=["trip_id", "date", "entry_station", "exit_station", "passenger_type", "duration_min", "peak_hour", "fare"])
df.to_csv("Travels_info.csv",index=False)
        
```
