```python
import pandas as pd 
import numpy as np 
df=pd.read_csv("Travels_info.csv")
```

```python
print("Total Trips:", len(df))
print("Total Revenue:", df['fare'].sum())
print("Average Fare:", df['fare'].mean())
print("Average Duration (min):", df['duration_min'].mean())
```

```python
print(df.groupby("passenger_type")["fare"].sum())
print(df["passenger_type"].value_counts(normalize=True) * 100)#% of passengers
```

```python
print(df["peak_hour"].value_counts(normalize=True) * 100)
print(df.groupby("peak_hour")["fare"].sum())
```

```python
print(df["entry_station"].value_counts())
print(df["exit_station"].value_counts())
```

```python
daily_revenue = df.groupby("date")["fare"].sum()
df["date"] = pd.to_datetime(df["date"])
monthly_revenue = df.groupby(df["date"].dt.to_period("M"))["fare"].sum()
print(daily_revenue)
print(monthly_revenue)
```

```python
print(df.loc[df["duration_min"].idxmax()])
print(df.loc[df["duration_min"].idxmin()])
```

```python
print(df["fare"].describe())
```

```python
print(df.groupby(["entry_station", "exit_station"])["fare"].sum().sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt

ptype_revenue = df.groupby("passenger_type")["fare"].sum()

plt.figure(figsize=(6,6))
plt.pie(ptype_revenue, labels=ptype_revenue.index, autopct="%1.1f%%", startangle=90)
plt.title("Revenue Contribution by Passenger Type")
plt.show()
```

```python
ptype_trips = df["passenger_type"].value_counts()

plt.figure(figsize=(6,6))
ptype_trips.plot(kind="bar", color=["blue","green","orange"])
plt.title("Number of Trips by Passenger Type")
plt.xlabel("Passenger Type")
plt.ylabel("Trips")
plt.show()
```

```python
peak_counts = df["peak_hour"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(peak_counts, labels=["Non-Peak","Peak"], autopct="%1.1f%%", colors=["lightblue","red"])
plt.title("Peak vs Non-Peak Trips")
plt.show()
```

```python
entry_counts = df["entry_station"].value_counts().head(10)

plt.figure(figsize=(12,6))
entry_counts.plot(kind="bar", color="skyblue")
plt.title("Top 10 Busiest Entry Stations")
plt.xlabel("Station")
plt.ylabel("Number of Trips")
plt.xticks(rotation=45)
plt.show()
```

```python
daily_revenue = df.groupby("date")["fare"].sum()

plt.figure(figsize=(12,6))
daily_revenue.plot(color="purple")
plt.title("Daily Revenue Trend")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.hist(df["duration_min"], bins=30, color="orange", edgecolor="black")
plt.title("Distribution of Trip Durations")
plt.xlabel("Duration (minutes)")
plt.ylabel("Frequency")
plt.show()
```

```python
routes = df.groupby(["entry_station", "exit_station"])["fare"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))
routes.plot(kind="bar", color="green")
plt.title("Top 10 Routes by Revenue")
plt.ylabel("Revenue")
plt.xlabel("Route (Entry → Exit)")
plt.xticks(rotation=75)
plt.show()
```
